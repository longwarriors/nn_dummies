# Hook Function钩子函数
回调函数

```python
class Oneday:
    def __init__(self, name):
        self.name = name
        self.hook = None
        
    def register_mouse_press_hook(self, hook_fn):
        self.hook = hook_fn
        
    def on_left_kry_down(self):
        print("Start from here:")
        if self.hook is not None:
            self.hook(self.name)
        else:
            print("No hooked, there is nothing to monitoring!")
        print("End here.")

#钩子函数        
def do(name):
    print(f"Hello, {name}")
    
today = Oneday('Monday')
today.register_mouse_press_hook(do)
today.on_left_kry_down()
```

```python
'''实现一个通用往队列中插入内容的功能
需要在插入队列前，对数据进行筛选 input_filter_fn
插入队列 insert_queue
'''
class ContentStash:
    '''pipeline:
    1. input_filter
    2. insert_queue'''
    def __init__(self):
        self.input_filter_fn = None
        self.broker = []
    
    # 留钩子    
    def register_input_filter_hook(self, input_filter_fn):
        self.input_filter_fn = input_filter_fn
        
    def insert_queue(self, content):
        self.broker.append(content)
        
    def input_pipeline(self, content, use=False):
        if not use:
            return 
        if self.input_filter_fn:
            _filter = self.input_filter_fn(content)
            if not _filter:
                self.insert_queue(content)
                
def input_filter_hook(content):
    if content.get('time') is None:
        return 
    else:
        return content
    
    
content = {'filename': 'test.jpg', 
           'b64_file': "#test", 
           'data': {"result": "cat", "probability": 0.9}
           }
content_stash = ContentStash('audit', work_dir='')
content_stash.register_input_filter_hook(input_filter_hook) # 挂钩子
content_stash.input_pipeline(content)
```
